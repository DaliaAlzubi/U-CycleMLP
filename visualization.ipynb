```python
import pandas as pd

acdc_ucycle_mlp_each_class = 'ACDC_metrics_for_each_class_ucycle_mlp.csv'
acdc_levit_unet_each_class = 'ACDC_metrics_for_each_class_levit_unet.csv'
acdc_swin_unet_each_class = 'ACDC_metrics_for_each_class_swin_unet.csv'
acdc_trans_unet_each_class = 'ACDC_metrics_for_each_class_trans_unet.csv'

acdc_ucycle_mlp_each_class_df = pd.read_csv('ACDC_metrics_for_each_class_ucycle_mlp.csv')
acdc_levit_unet_each_class_df = pd.read_csv('ACDC_metrics_for_each_class_levit_unet.csv')
acdc_swin_unet_each_class_df = pd.read_csv('ACDC_metrics_for_each_class_swin_unet.csv')
acdc_trans_unet_each_class_df = pd.read_csv('ACDC_metrics_for_each_class_trans_unet.csv')
```

```python
acdc_ucycle_mlp_each_class_df   
```

```python
row_means = acdc_ucycle_mlp_each_class_df.iloc[:, 1:].mean(axis=1)

# Add the row means as a new column in the DataFrame
acdc_ucycle_mlp_each_class_df['row_mean'] = row_means

# Display the updated DataFrame with the new column
print(acdc_ucycle_mlp_each_class_df)
```

```python
row_means = acdc_levit_unet_each_class_df.iloc[:, 1:].mean(axis=1)

# Add the row means as a new column in the DataFrame
acdc_levit_unet_each_class_df['row_mean'] = row_means

# Display the updated DataFrame with the new column
print(acdc_levit_unet_each_class_df)
```

```python
row_means = acdc_swin_unet_each_class_df.iloc[:, 1:].mean(axis=1)

# Add the row means as a new column in the DataFrame
acdc_swin_unet_each_class_df['row_mean'] = row_means

# Display the updated DataFrame with the new column
print(acdc_swin_unet_each_class_df)
```

```python
row_means = acdc_trans_unet_each_class_df.iloc[:, 1:].mean(axis=1)

# Add the row means as a new column in the DataFrame
acdc_trans_unet_each_class_df['row_mean'] = row_means

# Display the updated DataFrame with the new column
print(acdc_trans_unet_each_class_df)
```

```python
import plotly.express as px
import plotly.graph_objects as go

metrics = ['Dice-Score', 'F1-Score', 'IoU-Score',]

fig = go.Figure()
fig.add_trace(go.Bar(
    x=metrics,
    y=acdc_ucycle_mlp_each_class_df['row_mean'].iloc[[0,3,2]],
    name='ucycle_mlp',
    marker_color='indianred'
))
fig.add_trace(go.Bar(
    x=metrics,
    y=acdc_levit_unet_each_class_df['row_mean'].iloc[[0,3,2]],
    name='levit_unet',
    marker_color='lightsalmon'
))
fig.add_trace(go.Bar(
    x=metrics,
    y=acdc_swin_unet_each_class_df['row_mean'].iloc[[0,3,2]],
    name='swin_unet',
    marker_color='lightblue'
))
fig.add_trace(go.Bar(
    x=metrics,
    y=acdc_trans_unet_each_class_df['row_mean'].iloc[[0,3,2]],
    name='trans_unet',
    marker_color='lightgreen'
))

# Here we modify the tickangle of the xaxis, resulting in rotated labels.
fig.update_layout(barmode='group')#, xaxis_tickangle=-45)
fig.show()
```
